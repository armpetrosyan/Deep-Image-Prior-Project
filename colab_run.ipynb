## 5. Save results to Google Drive

The repo's `results/` folder becomes a link to `MyDrive/DIP_results`, so every script saves to Drive.

```python
DRIVE_RESULTS = "/content/drive/MyDrive/DIP_results"
LOCAL_RESULTS = os.path.join(REPO_DIR, "results")

os.makedirs(DRIVE_RESULTS, exist_ok=True)

if os.path.islink(LOCAL_RESULTS):
    print("results/ already links to", os.readlink(LOCAL_RESULTS))
elif os.path.exists(LOCAL_RESULTS):
    # Never delete a real folder: it may hold results that are not on Drive yet.
    print("WARNING: results/ is a real folder, so it is NOT saved to Drive.")
    print("Copy anything you need from it to Drive, then delete it and run this cell again.")
else:
    os.symlink(DRIVE_RESULTS, LOCAL_RESULTS)
    print("results/ ->", DRIVE_RESULTS)
```
